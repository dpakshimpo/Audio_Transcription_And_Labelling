```python
import whisperx
import torch
import gc
import pandas as pd
import os
import re
from google.colab import files

# -----------------------------
# CONFIG
# -----------------------------
audio_file = part1_file   # already uploaded earlier
output_txt = "part1_clean_speaker_transcript.txt"

# Use "large-v3" for better accuracy.
# If CUDA memory error happens, change this to "medium".
model_name = "large-v3"

device = "cuda" if torch.cuda.is_available() else "cpu"
compute_type = "float16" if device == "cuda" else "int8"
batch_size = 8 if device == "cuda" else 1

print("Audio file:", audio_file)
print("Device:", device)
print("Model:", model_name)
print("Compute type:", compute_type)
print("Batch size:", batch_size)

# -----------------------------
# HELPER FUNCTIONS
# -----------------------------
def clean_text(text):
    """Normalize spacing but preserve wording."""
    text = text.strip()
    text = re.sub(r"\s+", " ", text)
    return text

def speaker_display_name(speaker_id, speaker_map):
    """
    Convert pyannote labels like SPEAKER_00 / SPEAKER_01
    into Speaker 1 / Speaker 2.
    """
    if speaker_id not in speaker_map:
        speaker_map[speaker_id] = f"Speaker {len(speaker_map) + 1}"
    return speaker_map[speaker_id]

def merge_segments_by_speaker(segments):
    """
    Merge consecutive transcript segments from the same speaker.
    Returns blocks like:
    Speaker 1: text...
    Speaker 2: text...
    """
    speaker_map = {}
    blocks = []

    current_speaker = None
    current_text_parts = []

    for seg in segments:
        raw_speaker = seg.get("speaker", "UNKNOWN_SPEAKER")
        speaker = speaker_display_name(raw_speaker, speaker_map)
        text = clean_text(seg.get("text", ""))

        if not text:
            continue

        if speaker == current_speaker:
            current_text_parts.append(text)
        else:
            if current_speaker is not None:
                blocks.append((current_speaker, " ".join(current_text_parts)))

            current_speaker = speaker
            current_text_parts = [text]

    if current_speaker is not None:
        blocks.append((current_speaker, " ".join(current_text_parts)))

    return blocks, speaker_map

# -----------------------------
# 1. LOAD WHISPERX MODEL
# -----------------------------
print("\nLoading WhisperX ASR model...")

model = whisperx.load_model(
    model_name,
    device=device,
    compute_type=compute_type,
    language="en"
)

print("ASR model loaded.")

# -----------------------------
# 2. LOAD AUDIO
# -----------------------------
print("\nLoading audio...")

audio = whisperx.load_audio(audio_file)

print("Audio loaded.")

# -----------------------------
# 3. TRANSCRIBE
# -----------------------------
print("\nTranscribing full Part 1...")

asr_result = model.transcribe(
    audio,
    batch_size=batch_size,
    language="en"
)

print("Transcription complete.")
print("Transcript segments:", len(asr_result["segments"]))

# Free ASR model memory before alignment/diarization
del model
gc.collect()
if device == "cuda":
    torch.cuda.empty_cache()

# -----------------------------
# 4. ALIGN WORDS
# -----------------------------
print("\nLoading alignment model...")

align_model, metadata = whisperx.load_align_model(
    language_code=asr_result["language"],
    device=device
)

print("Aligning words...")

aligned_result = whisperx.align(
    asr_result["segments"],
    align_model,
    metadata,
    audio,
    device,
    return_char_alignments=False
)

print("Alignment complete.")

# Free alignment model memory
del align_model
gc.collect()
if device == "cuda":
    torch.cuda.empty_cache()

# -----------------------------
# 5. DIARIZATION
# -----------------------------
print("\nRunning diarization...")

diarization = pipeline(
    audio_file,
    min_speakers=2,
    max_speakers=2
)

print("Diarization complete.")

# -----------------------------
# 6. CONVERT PYANNOTE 4.x OUTPUT TO DATAFRAME
# -----------------------------
print("\nConverting diarization output to dataframe...")

annotation = diarization.speaker_diarization

rows = []
for turn, _, speaker in annotation.itertracks(yield_label=True):
    rows.append({
        "start": turn.start,
        "end": turn.end,
        "speaker": speaker
    })

diarize_df = pd.DataFrame(rows)

print("Diarization rows:", len(diarize_df))
print(diarize_df.head())

# -----------------------------
# 7. ASSIGN SPEAKERS TO WORDS/SEGMENTS
# -----------------------------
print("\nAssigning speakers to transcript...")

speaker_result = whisperx.assign_word_speakers(
    diarize_df,
    aligned_result
)

print("Speaker assignment complete.")

# -----------------------------
# 8. MERGE SAME-SPEAKER BLOCKS
# -----------------------------
print("\nCreating clean transcript without timestamps...")

blocks, speaker_map = merge_segments_by_speaker(speaker_result["segments"])

print("Speaker map:", speaker_map)
print("Transcript blocks:", len(blocks))

# -----------------------------
# 9. SAVE CLEAN TXT TRANSCRIPT
# -----------------------------
with open(output_txt, "w", encoding="utf-8") as f:
    f.write("Part 1 Transcript\n")
    f.write("=================\n\n")
    f.write("Note: Speaker labels are machine-generated. Please verify Speaker 1 / Speaker 2 manually.\n\n")

    for speaker, text in blocks:
        f.write(f"{speaker}: {text}\n\n")

print("Saved:", output_txt)

# -----------------------------
# 10. DOWNLOAD
# -----------------------------
files.download(output_txt)
```

```python
import os
import subprocess
from google.colab import files

# Your already-uploaded WAV variables
input_files = [
    part1_file,
    part2_file,
    part3_file
]

# Output consolidated WAV
combined_wav = "combined_parts_1_2_3.wav"

# Create ffmpeg concat list
concat_list = "wav_concat_list.txt"

with open(concat_list, "w", encoding="utf-8") as f:
    for audio_file in input_files:
        # ffmpeg concat list needs this format
        f.write(f"file '{os.path.abspath(audio_file)}'\n")

print("Concat list created:")
with open(concat_list, "r", encoding="utf-8") as f:
    print(f.read())

# Concatenate WAV files
# Re-encode to mono 16k PCM WAV to keep the final file clean and consistent
subprocess.run([
    "ffmpeg",
    "-y",
    "-f", "concat",
    "-safe", "0",
    "-i", concat_list,
    "-ar", "16000",
    "-ac", "1",
    "-c:a", "pcm_s16le",
    combined_wav
], check=True)

print("Combined WAV created:", combined_wav)
print("Size:", round(os.path.getsize(combined_wav) / (1024 * 1024), 2), "MB")

# Download the combined WAV
files.download(combined_wav)
```
